## Objective 3 — Forward and Backward Using While Loop


```python
# Name: Smarak Padhi
# Regd no: 25E116A08

s = input("Enter a string: ")
print("Forward Direction:")
i = 0
while i < len(s):
    print(s[i], end=" ")
    i += 1
print("\nBackward Direction:")
i = len(s) - 1
while i >= 0:
    print(s[i], end=" ")
    i -= 1
```

## Objective 4 — Basic String Encryption


```python
# Name: Smarak Padhi
# Regd no: 25E116A08

text = input("Enter a string: ")
encrypted = ""
for ch in text:
    if ch.isalpha():
        if ch.isupper():
            encrypted += chr((ord(ch) - 65 + 3) % 26 + 65)
        else:
            encrypted += chr((ord(ch) - 97 + 3) % 26 + 97)
    else:
        encrypted += ch
print("Encrypted String:", encrypted)
```

## Objective 5(b) — Eliminate Duplicate Values


```python
# Name: Smarak Padhi
# Regd no: 25E116A08

def remove_duplicates(lst):
    result = []
    for item in lst:
        if item not in result:
            result.append(item)
    return result

numbers = []
print("Enter 10 integers:")
for i in range(10):
    n = int(input("Enter number: "))
    numbers.append(n)
print("Original List:", numbers)
print("List after removing duplicates:", remove_duplicates(numbers))
```

## Objective 9 — 4 × 4 Matrix


```python
# Name: Smarak Padhi
# Regd no: 25E116A08

import random
matrix = []
for i in range(4):
    row = []
    for j in range(4):
        row.append(random.randint(0, 1))
    matrix.append(row)

print("Matrix:")
for row in matrix:
    print(*row)

row_counts = []
for row in matrix:
    row_counts.append(sum(row))

column_counts = []
for j in range(4):
    count = 0
    for i in range(4):
        count += matrix[i][j]
    column_counts.append(count)

max_row = row_counts.index(max(row_counts))
max_column = column_counts.index(max(column_counts))
print("The largest row index:", max_row)
print("The largest column index:", max_column)
```

## Objective 13 — Reverse First n Characters in a File
